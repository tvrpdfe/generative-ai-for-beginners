```python
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
```

```python
# 自动下载模型到本地缓存
model = SentenceTransformer('BAAI/bge-small-zh-v1.5')  # 多语言，性能优秀

texts = ["今天我们将学习人工智能", "机器学习是人工智能的分支"]
embeddings = model.encode(texts)

print(embeddings.shape)
print(embeddings.tolist())
print(embeddings)
```

```python
# compare several words
automobile_embedding    = model.encode('汽车').tolist()
vehicle_embedding       = model.encode('交通工具').tolist()
car_embedding           = model.encode('轿车').tolist()
dinosaur_embedding      = model.encode('恐龙').tolist()
stick_embedding         = model.encode('火柴').tolist()

# comparing cosine similarity, automobiles vs automobiles should be 1.0, i.e exactly the same, while automobiles vs dinosaurs should be between 0 and 1, i.e. not the same
print(cosine_similarity(automobile_embedding, automobile_embedding))
print(cosine_similarity(automobile_embedding, vehicle_embedding))
print(cosine_similarity(automobile_embedding, car_embedding))
print(cosine_similarity(automobile_embedding, dinosaur_embedding))
print(cosine_similarity(automobile_embedding, stick_embedding))
print(cosine_similarity(vehicle_embedding, car_embedding))
```
